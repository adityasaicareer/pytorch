```python
import torch
from torch import nn
import numpy as np
import pandas as pd
```

```python
device='mps' if torch.backends.mps.is_available() else 'cpu'
x=np.arange(0,1,0.01)
y=0.8*(x**2)
limit=int(0.8*len(x))
x_train=torch.from_numpy(x[:int(0.8*len(x))]).type(torch.float32).to(device)
y_train=torch.from_numpy(y[:int(0.8*len(x))]).type(torch.float32).to(device)
x_test=torch.from_numpy(x[int(0.8*len(x)):]).type(torch.float32).to(device)
y_test=torch.from_numpy(y[int(0.8*len(x)):]).type(torch.float32).to(device)
```

```python
class regression(nn.Module):
    def __init__(self):
        super().__init__()
        self.weight=nn.Parameter(torch.rand(1,requires_grad=True,dtype=torch.float32))
        print(self.weight)
    def forward(self,x):
        return self.weight*(x**2)
    



```

```python
torch.manual_seed(42)
model=regression()
loss_fn=nn.L1Loss()
optimizer=torch.optim.SGD(params=model.parameters(),lr=0.01)
model.state_dict()
list(model.parameters())
```

```python
epochs=100
for epoch in range(epochs):
    model.train()
    y_p=model(x_train)
    loss=loss_fn(y_p,y_train)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```
